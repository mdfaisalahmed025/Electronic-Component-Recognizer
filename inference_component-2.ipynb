```python
electronic_labels = [
    "resistor",
    "capacitor",
    "diode",
    "transistor",
    "ic chip",
    "relay",
    "switch",
    "transformer",
    "led",
    "connector",
    "potentiometer",
    "fuse",
    "inductor coil",
    "crystal oscillator",
    "heat sink",
    "breadboard",
    "jumper wire",
    "battery holder",
    "power supply module",
    "microcontroller board"
]

def recognize_image(image):
    # make sure image is RGB
    if image.mode != "RGB":
        image = image.convert("RGB")

    # convert to fastai PILImage and predict
    img = PILImage.create(image)
    pred, idx, probs = model.predict(img)

    # get labels from the learner (ensure it's a flat list)
    labels = model.dls.vocab
    if isinstance(labels[0], (list, tuple)):
        labels = labels[0]

    probs = list(map(float, probs))
    result = {lab: p for lab, p in zip(labels, probs)}

    # return sorted dict so Gradio shows highest first
    return dict(sorted(result.items(), key=lambda x: x[1], reverse=True))

```

```python
import gradio as gr

image = gr.Image(type="pil", height=192, width=192)
label = gr.Label()

examples = [
    'test-images/Breadboard.jpg',
    'test-images/ic-chip.png',
    'test-images/battery-holder.jpg',
    'test-images/diode.png',
    'test-images/transistor.jpg',
    'test-images/connector.webp',

]

iface = gr.Interface(
    fn=recognize_image,
    inputs=image,
    outputs=label,
    examples=examples
)

iface.launch(inline=False, share=True)
```
